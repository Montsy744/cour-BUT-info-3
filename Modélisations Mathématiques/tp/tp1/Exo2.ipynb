```python
#Import Bibliothèque

import numpy as np
import matplotlib.pyplot as plt
```

**Q2. Commençons par visualiser une loi normale.**

**Q2.1.** Définition de la fonction de densité

```python
def densité(mu, sigma, x):
    """Calcule la densité de probabilité de la loi N(mu, sigma) pour une valeur/array x."""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma)**2)
```

**Q2.2.** Génération du vecteur $x$

```python
x = np.linspace(0, 80, 1000)
```

**Q2.3.** Tracé de la densité $\mathcal{N}(35, 10)$

plt.figure(figsize=(8, 5))
plt.plot(x, densité(35, 10, x), label="$\mathcal{N}(35, 10)$", color="blue")
plt.title("Densité de la loi normale $\mathcal{N}(35, 10)$")
plt.xlabel("x")
plt.ylabel("Densité de probabilité")
plt.legend(loc="upper right")
plt.grid(True, alpha=0.3)
plt.show()

**Q3. Compréhension des paramètres de la loi normale**

**Q3.1.** Impact de l'écart-type $\sigma$ (Espérance fixe $\mu = 35$)

```python
plt.figure(figsize=(9, 5))
plt.plot(x, densité(35, 10, x), label="$\mathcal{N}(35, 10)$ - $\sigma=10$")
plt.plot(x, densité(35, 20, x), label="$\mathcal{N}(35, 20)$ - $\sigma=20$")
plt.plot(x, densité(35, 5, x), label="$\mathcal{N}(35, 5)$ - $\sigma=5$")

plt.title("Impact de l'écart-type $\sigma$ sur la loi normale")
plt.xlabel("x")
plt.ylabel("Densité")
plt.legend(loc="upper right")
plt.grid(True, alpha=0.3)
plt.show()
```

*Constat : L'écart-type $\sigma$ mesure la dispersion. Plus $\sigma$ est petit, plus la courbe est étroite et élevée autour de la moyenne. Plus $\sigma$ est grand, plus la courbe est étalée et aplatie.*

**Q3.2.** Impact de l'espérance $\mu$ (Écart-type fixe $\sigma = 10$)

```python
plt.figure(figsize=(9, 5))
plt.plot(x, densité(35, 10, x), label="$\mathcal{N}(35, 10)$ - $\mu=35$")
plt.plot(x, densité(45, 10, x), label="$\mathcal{N}(45, 10)$ - $\mu=45$")
plt.plot(x, densité(55, 10, x), label="$\mathcal{N}(55, 10)$ - $\mu=55$")

plt.title("Impact de l'espérance $\mu$ sur la loi normale")
plt.xlabel("x")
plt.ylabel("Densité")
plt.legend(loc="upper right")
plt.grid(True, alpha=0.3)
plt.show()
```

*Constat : L'espérance $\mu$ est un paramètre de position. Modifier $\mu$ déplace la courbe horizontalement le long de l'axe des abscisses sans changer sa forme ni sa hauteur*

**Q3.3.** Estimation des probabilités pour $X \sim \mathcal{N}(35, 10)$

Par symétrie et propriété de la loi normale (règle des $68-95-99.7$) :

- $P[X < 35] = \mathbf{0.50}$ (50% de l'aire sous la courbe, la distribution étant symétrique par rapport à la moyenne $\mu = 35$).
- $P[X < 45] \approx \mathbf{0.8413}$ (car $45 = \mu + \sigma$ ; l'aire entre $\mu-\sigma$ et $\mu+\sigma$ vaut environ $68.3\%$, d'où $50\% + 34.15\% \approx 84.15\%$).
- $P[35 < X < 45] = P[X < 45] - P[X < 35] \approx 0.8413 - 0.50 = \mathbf{0.3413}$ (34.13%).$P[X > 45] = 1 - P[X < 45] \approx 1 - 0.8413 = \mathbf{0.1587}$ (15.87%).

**Q4. Simulation d'échantillons et histogrammes**

**Q4.1** & **Q4.2.** Échantillon de taille $N = 100$ et tracé

```python
rng = np.random.default_rng(seed=42)

ages100 = rng.normal(loc=35, scale=10, size=100)

plt.figure(figsize=(8, 5))
plt.hist(ages100, density=True, alpha=0.25, color="blue", edgecolor="black", label="Échantillon N=100")
plt.plot(x, densité(35, 10, x), color="red", linewidth=2, label="Densité théorique $\mathcal{N}(35, 10)$")
plt.title("Histogramme d'un échantillon de 100 valeurs vs Densité théorique")
plt.xlabel("Âge")
plt.ylabel("Densité")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

**Q4.3** & **Q4.4.** Impact de la taille de l'échantillon ($N = 10, 10^5, 10^6$)

```python
ages10 = rng.normal(35, 10, 10)
ages10K = rng.normal(35, 10, 100000)
ages1M = rng.normal(35, 10, 1000000)

plt.figure(figsize=(10, 6))
plt.hist(ages10, density=True, alpha=0.3, label="N = 10", bins=10)
plt.hist(ages10K, density=True, alpha=0.3, label="N = 100 000", bins=100)
plt.hist(ages1M, density=True, alpha=0.3, label="N = 1 000 000", bins=200)

plt.plot(x, densité(35, 10, x), color="black", linewidth=2, linestyle="--", label="Densité théorique")
plt.title("Impact de la taille de l'échantillon sur l'estimation de la densité")
plt.xlabel("Âge")
plt.ylabel("Densité")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

*Constat : En vertu de la Loi des Grands Nombres, plus la taille $N$ de l'échantillon augmente, plus l'histogramme empirique converge parfaitement vers la courbe de densité théorique.*

**Q4.5.** Impact du nombre d'intervalles (bins) pour ages10K

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

bins_values = [10, 100, 1000]

for ax, b in zip(axes, bins_values):
    ax.hist(ages10K, bins=b, density=True, alpha=0.5, color="teal", edgecolor="none")
    ax.plot(x, densité(35, 10, x), color="red", linewidth=1.5)
    ax.set_title(f"bins = {b}")
    ax.set_xlabel("Âge")
    ax.grid(True, alpha=0.3)

axes[0].set_ylabel("Densité")
plt.suptitle("Impact du paramètre `bins` sur l'histogramme (N = 100 000)", fontsize=14)
plt.tight_layout()
plt.show()
```

Constat sur le paramètre bins :

    bins = 10 (sous-lissage) : Intervalles trop larges, perte de précision spatiale de la forme de la courbe.

    bins = 100 (équilibre idéal) : La forme de la loi normale est fidèlement restituée.

    bins = 1000 (sur-lissage / bruit) : Intervalles trop étroits, l'histogramme devient très "bruité" car chaque classe contient peu de données.
