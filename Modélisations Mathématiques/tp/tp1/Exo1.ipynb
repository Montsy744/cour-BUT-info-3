**Q2 : Travaillons d’abord sur une structure d’une seule dimension ! Nous disposons d’une liste contenant les revenus mensuels de vos clients et leurs noms dans une autre. Il faut comprendre que Ava gagne 1800C alors que Bob gagne 1500C mensuels. Notez que pratiquement toutes les questions du TPs sont à faire sans écrire la moindre boucle (sauf mention explicite) !**

**Q2.1.** Créez un array nommé revenus à partir de liste_revenus et affichez son contenu. Faites de même pour
les noms.

```python
liste_revenus = [1800, 1500, 2200, 3000, 2172, 5000, 1400, 1200, 1100, 1300]
liste_noms = ['Ava', 'Bob', 'Cao', 'Dan', 'Eli', 'Fay', 'Guy', 'Hue', 'Ian', 'Joy']

revenus = np.array(liste_revenus)
noms = np.array(liste_noms)

print("Array des revenus :", revenus)
print("Array des noms :", noms)
```

**Q2.2.** Créez un array hauts_revenus contenant uniquement les revenus supérieurs à 2000C, via une indexation
booléenne ou via la fonction where(...) de numpy. Affichez son contenu.

```python
# Renvoie les indices des éléments qui vérifient la condition
hauts_revenus = np.where(revenus > 2000)

print(hauts_revenus)
```

**Q2.3.** Affichez les noms des personnes gagnant plus de 2000C mensuels, à partir de la fonction where(...).

```python
noms_hauts_revenus = noms[hauts_revenus]

print("Personnes gagnant plus de 2000 € :", noms_hauts_revenus)
```

**Q2.4.** Calculez la moyenne des revenus de tous vos clients, via la méthode mean(). Affichez en arrondissant grâce
à une f-string.

```python
moyenne_revenus = revenus.mean()

print(f"La moyenne des revenus est de {moyenne_revenus:.2f} €")
```

**Q2.5.** Calculez et affichez en arrondissant (en KC) la somme des revenus annuels de tous vos clients.

```python
somme_annuelle_kc = (revenus * 12).sum() / 1000

print(f"Somme des revenus annuels de tous les clients : {somme_annuelle_kc:.2f} k€")
```

**Q2.6.** Le client gagnant 1100C a reçu une augmentation de 150C. De même, Ava a été augmenté de 100C. Ré-
percutez ces hausses dans votre structure. Veillez à n’utiliser que les informations à votre disposition pour effectuer
chacune des opérations : à partir de la valeur elle-même et l’indexation booléenne d’un côté, à partir du nom à associer
avec la fonction where(...) de l’autre.

```python
revenus[revenus == 1100] += 150

index_ava = np.where(noms == 'Ava')
revenus[index_ava] += 100

print("Array revenus mis à jour :", revenus)
```

**Q2.7.** Affichez les noms et revenus de chacun des clients à raison d’un client par ligne, au moyen d’une boucle.

```python
for nom, revenu in zip(noms, revenus):
    print(f"{nom} : {revenu} €")
```

**Q3. Travaillons désormais en deux dimensions. Nous disposons toujours d’une liste de clients qui incluent leur revenu, leur âge et leur nombre d’enfants à charge. À chaque ligne correspond un client : liste_clients contient ses informations tandis que liste_noms contient son nom à l’indice correspondant.**

```python
clients = np.array([
    [1800, 21, 0],
    [1500, 54, 2],
    [2200, 28, 3],
    [3000, 37, 1],
    [2172, 37, 2],
    [5000, 32, 0],
    [1400, 23, 0],
    [1200, 25, 1],
    [1100, 19, 0],
    [1300, 31, 2]
])

noms = np.array(['Ava', 'Bob', 'Cao', 'Dan', 'Eli', 'Fay', 'Guy', 'Hue', 'Ian', 'Joy'])
```

**Q3.1.** Cao souhaiterait contracter un prêt immobilier. Affichez les informations sur Cao, via la fonction index(...).
Calculez ses mensualités maximales possibles, en sachant que le taux d’endettement maximum est de 35% (il ne pourra
donc pas rembourser plus de 35% de son revenu par mois).

```python
index_cao = liste_noms.index('Cao')

infos_cao = clients[index_cao]
print("Informations sur Cao [Revenu, Âge, Enfants] :", infos_cao)

revenu_cao = infos_cao[0]
mensualite_max = revenu_cao * 0.35

print(f"Mensualité maximale possible pour Cao : {mensualite_max:.2f} €")
```

**Q3.2.** Kim est une nouvelle cliente qui vient d’arriver : kim = [1900, 31, 1]. Ajoutez ces informations à vos
données via la fonction append(...) ou vstack(...). Affichez à nouveau les informations (et vérifiez les dimensions
du tableau obtenu).

```python
kim_infos = [1900, 31, 1]
kim_nom = 'Kim'

clients = np.vstack([clients, kim_infos])

noms = np.append(noms, kim_nom)

print("Dimensions de la nouvelle matrice clients :", clients.shape)
print("\nDernière ligne ajoutée (Kim) :", clients[-1])
```

**Q3.3.** Stockez enfin l’ensemble des informations sur les âges de notre clientèle, correspondant à la colonne d’indice
1, dans une variable clients_ages et affichez-la. Les données sont-elles encore sous forme de colonne ?

```python
clients_ages = clients[:, 1]

print("Contenu de clients_ages :", clients_ages)
print("Forme (shape) de clients_ages :", clients_ages.shape)
```
